```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1) Load tips dataset
tips_dataset = sns.load_dataset('tips')

print("Tips Dataset:")
print(tips_dataset.head())


# 2) Histogram for tip with day
sns.set_style('whitegrid')

sns.histplot(
    x=tips_dataset.tip,
    hue=tips_dataset.day,
    element='step'
)

plt.show()


# 3) Displot for day with sex
sns.displot(
    x=tips_dataset.day,
    hue=tips_dataset.sex,
    multiple='dodge',
    shrink=0.8,
    palette='Set1',
    hue_order=['Female', 'Male']
)

plt.show()


# 4) ECDF plot for tip with time
sns.set_style('darkgrid')
sns.set(font_scale=1.25)

sns.ecdfplot(
    x=tips_dataset.tip,
    hue=tips_dataset.time,
    palette='summer',
    linewidth=3
)

plt.show()
```
